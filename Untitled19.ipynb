```python
import pandas as pd
import numpy as np
from string import punctuation
from nltk.corpus import stopwords
import nltk
from nltk.stem import WordNetLemmatizer
import pickle
from nltk.stem import PorterStemmer
import os
```

```python
from zeugma.embeddings import EmbeddingTransformer#loading glove embedding vector  class
from sklearn.cluster import KMeans
from sklearn.feature_selection import SelectKBest
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB #loading naive vayes
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA #loaidng PCA class
from sklearn.feature_extraction.text import TfidfVectorizer #loading tfidf vector
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
from sklearn.decomposition import TruncatedSVD
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
```

```python
#define object to remove stop words and other text processing
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
ps = PorterStemmer()
```

```python
#define function to clean text by removing stop words and other special symbols
def cleanText(doc):
    tokens = doc.split()
    table = str.maketrans('', '', punctuation)
    tokens = [w.translate(table) for w in tokens]
    tokens = [word for word in tokens if word.isalpha()]
    tokens = [w for w in tokens if not w in stop_words]
    tokens = [word for word in tokens if len(word) > 1]
    tokens = [ps.stem(token) for token in tokens]
    tokens = [lemmatizer.lemmatize(token) for token in tokens]
    tokens = ' '.join(tokens)
    return tokens
```

```python
#function to get class label from given news category
def getLabel(category):
    label = -1
    for i in range(len(labels)):
        if labels[i] == category:
            label = i
            break
    return label
```

```python
#loading & displaying dataset values
dataset = pd.read_csv("bbc-news-data.csv", sep="\t")
labels = np.unique(dataset['category']).ravel()
dataset
```

```python
#preprocess & clean each news data
if os.path.exists("X.npy"):
    X = np.load("X.npy")
    Y = np.load("Y.npy")
else:
    content = dataset['content'].ravel()
    news_category = dataset['category'].ravel()
    X = []
    Y = []
    for i in range(len(content)):#looping all news content 
        news = content[i]
        news = news.strip("\n").strip().lower()
        label = getLabel(news_category[i])
        news = cleanText(news)#clean news
        X.append(news)
        Y.append(label)
    X = np.asarray(X)
    Y = np.asarray(Y)
    np.save("X", X)
    np.save("Y", Y)
#visualizing class labels count found in dataset
names, count = np.unique(Y, return_counts = True)
height = count
bars = labels
y_pos = np.arange(len(bars))
plt.figure(figsize = (4, 3)) 
plt.bar(y_pos, height)
plt.xticks(y_pos, bars)
plt.xlabel("Different News Category Label Graph")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()    
```

```python
#define global variables to calculate and store accuracy and other metrics
precision = []
recall = []
fscore = []
accuracy = []
```

```python
#function to calculate various metrics such as accuracy, precision etc
def calculateMetrics(algorithm, predict, testY):
    p = precision_score(testY, predict,average='macro') * 100
    r = recall_score(testY, predict,average='macro') * 100
    f = f1_score(testY, predict,average='macro') * 100
    a = accuracy_score(testY,predict)*100     
    print(algorithm+' Accuracy  : '+str(a))
    print(algorithm+' Precision   : '+str(p))
    print(algorithm+' Recall      : '+str(r))
    print(algorithm+' FMeasure    : '+str(f))    
    accuracy.append(a)
    precision.append(p)
    recall.append(r)
    fscore.append(f)
    conf_matrix = confusion_matrix(testY, predict) 
    plt.figure(figsize =(4, 3)) 
    ax = sns.heatmap(conf_matrix, xticklabels = labels, yticklabels = labels, annot = True, cmap="viridis" ,fmt ="g");
    ax.set_ylim([0,len(labels)])
    plt.title(algorithm+" Confusion matrix") 
    plt.ylabel('True class') 
    plt.xlabel('Predicted class') 
    plt.show()
```

```python
#convert news data into glove vector and then apply clustering algorithm to reduce features set
if os.path.exists("glove.npy") == False:
    glove = EmbeddingTransformer('glove') #convert X news data into glove vector
    glove_X = glove.transform(X)
    np.save("glove", glove_X)
else:
    glove_X = np.load("glove.npy")
#shuffle dataset values    
indices = np.arange(X.shape[0])
np.random.shuffle(indices)
glove_X = glove_X[indices]
Y1 = Y[indices]        
#normalize training data
scaler = StandardScaler()
glove_X = scaler.fit_transform(glove_X)
#apply clustering algorithm to reduce features
kmeans = KMeans(n_clusters=2)
kmeans.fit(glove_X)
cluster_labels = kmeans.labels_
selector = SelectKBest(k=24)
selector.fit(glove_X, cluster_labels)
selected_features = selector.get_support()
glove_X = glove_X[:,selected_features] #extract selected features
X_train, X_test, y_train, y_test = train_test_split(glove_X, Y1, test_size=0.2) #split dataset into train and test
#train Naive Bayes algorithm on glove vector and calculate accuracy
nb = GaussianNB()
nb.fit(X_train, y_train)
predict = nb.predict(X_test)
calculateMetrics("Propose rFR", y_test, predict)
```

```python
#train PCA on X news data
tfidf_vectorizer = TfidfVectorizer(max_features=50)
XX = tfidf_vectorizer.fit_transform(X).toarray()
indices = np.arange(XX.shape[0])
np.random.shuffle(indices)
XX = XX[indices]
Y1 = Y[indices] 
#reducing features using OCA
pca = PCA(n_components=30)
pca_X = pca.fit_transform(XX)
#training Naive Bayes using PCA features
X_train, X_test, y_train, y_test = train_test_split(pca_X, Y1, test_size=0.2) #split dataset into train and test
nb = GaussianNB(var_smoothing=0.9)#training naive bayes on PCA features
nb.fit(X_train, y_train)
predict = nb.predict(X_test)
calculateMetrics("PCA", y_test, predict)
```

```python
#reducing features using LDA
lda = LDA(n_components=3)
lda_X = lda.fit_transform(XX, Y1)
X_train, X_test, y_train, y_test = train_test_split(lda_X, Y1, test_size=0.2) #split dataset into train and test
#training naive Bayes on LDA features
nb = GaussianNB()
nb.fit(X_train, y_train)
predict = nb.predict(X_test)
calculateMetrics("LDA", y_test, predict)
```

```python
#reducing features using LSI
lsi = TruncatedSVD(n_components=30)
lsi_X = lsi.fit_transform(XX)
X_train, X_test, y_train, y_test = train_test_split(lsi_X, Y1, test_size=0.2) #split dataset into train and test
#training naive bayes on LSI reduced features
nb = GaussianNB()
nb.fit(X_train, y_train)
predict = nb.predict(X_test)
calculateMetrics("LSI", y_test, predict)
```
